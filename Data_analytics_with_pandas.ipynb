# 4. Propiedades loc, iloc, at e iat 

```python
import pandas as pd

pd.set_option("display.notebook_repr_html", False)
```

```python
#
# Carga del archivo ddesde un repo en GitHub
#
drivers = pd.read_csv(
    "https://raw.githubusercontent.com/jdvelasq/playground/master/datasets/drivers/drivers.csv",
    sep=",",
    thousands=None,
    decimal=".",
)

drivers.set_index("name", inplace=True)
drivers.head()
```

```python
#
# iloc recibe únicamente enteros, por lo que
# permite seleccionar elementos por posición.
#
drivers.iloc[2:5]
```

```python
drivers.loc[
    [
        "Joe Niemiec",
        "Rohit Bakshi",
        "Adis Cesir",
    ]
]
```

```python
# Selección de un grupo de filas y columnas 

drivers.iloc[2:5, [1, 3]]
```

```python
# Selección de un grupo de filas y columnas 

rows = [
        "Joe Niemiec",
        "Rohit Bakshi",
        "Adis Cesir",
    ]

columns = [
    "location",
    "certified",
    "wage-plan",
]

drivers.loc[rows, columns]
```

```python
# Selección de un registro en particular

drivers.iloc[1]
```

```python
# Selección de un registro en particular

drivers.loc["Jamie Engesser"]
```

```python
# Selección de una celda en particular por posición
## iat es la opción más eficiente cuando solo necesitas el valor de una celda específica.
drivers.iat[1, 2]
```

```python
# Selección de una celda en particular usando etiquetas 

drivers.at["Joe Niemiec", "location"]
```

```python
#Concatenacion DF1 y Df2

# Carga del dataset 1
#
dataset_1 = pd.read_csv("/tmp/dataset_1.csv")

#
# Carga del dataset 2
#
dataset_2 = pd.read_csv("/tmp/dataset_2.csv")

#
# Concat con valores por defecto
#
datasets_12 = pd.concat(
    objs=[
        dataset_1,
        dataset_2,
    ],
    #
    # axis= 0/index, 1/columns
    #
    axis=0,
    #
    # join={'inner', 'outer'}
    #
    join="outer",
    #
    # ignore_index=bool
    #
    ignore_index=False,
    #
    # sort=bool
    # Ordena el otro eje si el join es 'outer'.
    # Se ignora cuando es 'inner'
    #
    sort=False,
)

#
# Observe que hay numeros de fila repetidos (index)
# ignore_index=False
#
datasets_12
```

```python
#
# Se ignora el índice
#
datasets_12 = pd.concat(
    objs=[
        dataset_1,
        dataset_2,
    ],
    ignore_index=True,
)
datasets_12
```

```python
bonus = pd.read_csv("/tmp/bonus.csv")

#
# Se requiere que los datasets tengan los mismos
# indices en las filas.
#
dataset_full = pd.concat(
    [
        datasets_12,
        bonus,
    ],
    axis=1,
)
dataset_full
```

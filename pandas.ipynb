# REMEMBER
****When selecting subsets of data, square brackets [] are used.
# Ejemplo: Seleccionar toda la columna 'Edad'
subconjunto = df['Edad']

****Inside these brackets, you can use a single column/row label, a list of column/row labels, a slice of labels, a conditional expression or a colon.
-Single column label (Etiqueta única):
df['Ciudad']  # Devuelve una Serie
-List of column labels (Lista de etiquetas):
df[['Edad', 'Ciudad']]  # Devuelve un DataFrame
-Slice of labels (Rebanado de etiquetas - generalmente para filas):
df['Ana':'Pedro']  # Selecciona filas desde 'Ana' hasta 'Pedro' inclusivo
-Conditional expression (Expresión condicional / Filtro booleano):
df[df['Edad'] > 30]  # Selecciona filas donde la Edad es mayor a 30

****Select specific rows and/or columns using loc when using the row and column names.

# Ejemplo: Fila 'Luis', Columna 'Ciudad'
valor = df.loc['Luis', 'Ciudad']
# Ejemplo: Filas de 'Luis' a 'Marta', columnas 'Edad' y 'Ciudad'
subconjunto_loc = df.loc['Luis':'Marta', ['Edad', 'Ciudad']]

****Select specific rows and/or columns using iloc when using the positions in the table.
# Ejemplo: Fila en posición 1 (Luis), Columna en posición 0 (Edad)
# Recuerda que Python empieza a contar desde 0
valor_num = df.iloc[1, 0]

# Ejemplo: Las primeras 2 filas, y las últimas 2 columnas
subconjunto_iloc = df.iloc[0:2, -2:]

***** You can assign new values to a selection based on loc/iloc.

# Usando loc para modificar un valor específico
# Cambiamos la Ciudad de 'Pedro' a 'Cusco'
df.loc['Pedro', 'Ciudad'] = 'Cusco'

# Usando iloc para modificar un bloque
# Ponemos la edad de las dos primeras personas en 0 (solo como ejemplo)
df.iloc[0:2, 0] = 0

 # Wide to long format#

Starting again from the wide format table created in the previous section, we add a new index to the DataFrame with reset_index().

```python
no2_pivoted = no2.pivot(columns="location", values="value").reset_index()
no2_pivoted.head()
```

```python
# I want to collect all no2 air quality  measurements in a single column (long format).


no_2 = no2_pivoted.melt(id_vars="date.utc")

no_2


# The pandas.melt() method on a DataFrame converts the data table from wide format to long format. The column headers become the variable names in a newly created column.
```

```python
#parameters to pandas.melt() defined in more detail

no_2 = no2_pivoted.melt(
    id_vars="date.utc",
    value_vars=["BETR801", "FR04014", "London Westminster"],
    value_name="NO_2",
    var_name="id_location",
)
```

The additional parameters have the following effects:

value_vars defines which columns to melt together
value_name provides a custom column name for the values column instead of the default column name value
var_name provides a custom column name for the column collecting the column header names. Otherwise it takes the index name or a default variable
Hence, the arguments value_name and var_name are just user-defined names for the two generated columns. The columns to melt are defined by id_vars and value_vars.
